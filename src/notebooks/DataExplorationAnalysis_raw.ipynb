## Load Raw Data

Load CSV data from the configured raw data path

```python
# Get the raw data file path from config
raw_data_file = config.DATA['RAW_PATH_FILE']
print(f"Raw data file: {raw_data_file}")

# Check if file exists and load
if os.path.exists(raw_data_file):
    print(f"Loading data from: {raw_data_file}")
    df = pd.read_csv(raw_data_file)
    print(f"✓ Data loaded successfully!")
else:
    print(f"\nFile not found: {raw_data_file}")
    print("Please ensure the CSV file exists at the configured path.")
    df = None
```
